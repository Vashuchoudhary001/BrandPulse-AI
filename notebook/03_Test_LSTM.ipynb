```python

from pathlib import Path
import joblib

BASE_DIR = Path.cwd().parent

MODEL_PATH = BASE_DIR / "models" / "lstm_sentiment_model.keras"
TOKENIZER_PATH = BASE_DIR / "models" / "lstm_tokenizer.pkl"

print("Model file exists:", MODEL_PATH.exists())
print("Tokenizer file exists:", TOKENIZER_PATH.exists())

assert MODEL_PATH.exists(), "LSTM model file not found!"
assert TOKENIZER_PATH.exists(), "Tokenizer file not found!"

tokenizer = joblib.load(TOKENIZER_PATH)

print("\nTokenizer loaded successfully!")
print("Vocabulary size:", len(tokenizer.word_index))
print("Tokenizer settings:", tokenizer.get_config().keys())
```

```python
import tensorflow as tf

# lstm load
lstm_model = tf.keras.models.load_model(MODEL_PATH)

print("LSTM model loaded successfully!")
print("Input shape:", lstm_model.input_shape)
print("Output shape:", lstm_model.output_shape)

# architexture checking 
lstm_model.summary()
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np

sample_text = ["I love this airline, the service was amazing!"]

# Text to numbers 
sequence = tokenizer.texts_to_sequences(sample_text)

# Model  required input length 50  according padding
padded_sequence = pad_sequences(
    sequence,
    maxlen=50,
    padding="post",
    truncating="post"
)

# Sentiment scores prediction
prediction = lstm_model.predict(padded_sequence, verbose=0)

print("Prediction scores:", prediction)
print("Predicted class index:", np.argmax(prediction))
```
